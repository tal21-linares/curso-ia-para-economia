### **Ejercicio 1: Carga y Preparación Inicial de los Datos**

Primero, carguemos las librerías necesarias y el dataset. Nos enfocaremos en un subconjunto de las variables numéricas para mantener el taller manejable, pero el principio se aplica a todo el dataset.

```python
# Importación de librerías estándar
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Importación de herramientas de scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
```

Mejorar visualización de dataframes y gráficos

```python
# Que muestre todas las columnas
pd.options.display.max_columns = None
# En los dataframes, mostrar los float con dos decimales
pd.options.display.float_format = '{:,.2f}'.format

# Configuraciones para una mejor visualización
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

Cargar el dataset

```python
# Usamos el dataset Ames Housing desde su fuente original (requiere internet)
url = 'http://jse.amstat.org/v19n3/decock/AmesHousing.txt'
df = pd.read_csv(url, sep='\t')
df.head()
```

```python
# Para este taller, solo usaremos algunas columnas clave para simplificar.
df_sub = df[['Overall Qual', 'Gr Liv Area', 'SalePrice']].copy()
df_sub.columns = ['OverallQual', 'GrLivArea', 'SalePrice'] # Renombrar para facilidad
df_sub.head()
```

**Explicación de las variables del dataset reducido**

1. SalePrice: Precio de Venta.

Esta es la variable objetivo. Es el precio final por el cual se vendió la propiedad, medido en dólares estadounidenses.

2. OverallQual: Calidad General.

Es una variable ordinal que califica la calidad general del material y el acabado de la casa. Es una de las variables predictoras (X) más importantes.

Escala: Va de 1 a 10.

10: Muy Excelente

9: Excelente

...

2: Pobre

1: Muy Pobre

3. GrLivArea: Área Habitable sobre el Nivel del Suelo.

Es una variable numérica que mide el total de metros cuadrados de área habitable que está por encima del nivel del suelo. No incluye el área del sótano. Es una de las variables predictoras (X) más fuertes, ya que, lógicamente, casas más grandes tienden a ser más caras.

```python
df_sub.info()
```

### **Ejercicio 2: Dividir el conjunto de datos**

El método `.info()` muestra que no hay nulos en nuestro subconjunto. ¡Perfecto! Ahora, definamos nuestras variables `X` (predictoras) e `y` (objetivo) y dividamos los datos.

```python
X = df_sub[['OverallQual', 'GrLivArea']]
y = df_sub['SalePrice']
```

```python
# Dividir en entrenamiento y prueba (70/30)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### **Ejercicio 3: Modelo con Sobreajuste**

Vamos a crear un modelo muy complejo para ver qué tan mal puede generalizar.

**Crear un Modelo Polinómico de Grado 5**

Usa `Pipeline` para combinar `PolynomialFeatures` (grado 5), `StandardScaler` y `LinearRegression`. Entrénalo con los datos de entrenamiento.

```python
# Crear el pipeline para el modelo polinómico
modelo_polinomico = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('model', LinearRegression())
])
```

```python
# Entrenar el modelo
modelo_polinomico.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
#Predicciones
y_train_pred_poly = modelo_polinomico.predict(X_train)
y_test_pred_poly = modelo_polinomico.predict(X_test)

# Cálculo del error (RMSE)
rmse_train_poly = np.sqrt(mean_squared_error(y_train, y_train_pred_poly))
rmse_test_poly = np.sqrt(mean_squared_error(y_test, y_test_pred_poly))

print(f"RMSE en Entrenamiento (Polinómico): ${rmse_train_poly:,.2f}")
print(f"RMSE en Prueba (Polinómico): ${rmse_test_poly:,.2f}")
```

**Pregunta:** ¿Qué indica la diferencia que observas entre el error de entrenamiento y el de prueba? ¿Le recomendarías este modelo a la firma inmobiliaria? ¿Por qué?

La diferencia significativa entre el error de entrenamiento y el de prueba indica la presencia de sobreajuste. El modelo ha aprendido en exceso los patrones específicos del conjunto de entrenamiento, incluyendo el ruido, lo que reduce su capacidad de generalización.

Por esta razón, no se recomienda su implementación en un entorno real, ya que su desempeño fuera de muestra es inestable y poco fiable para la toma de decisiones.

### **Ejercicio 4: Aplicar Regularización**

Ahora, vamos a "curar" el sobreajuste. Usaremos los mismos `PolynomialFeatures` de grado 5, pero cambiaremos el modelo de regresión.

**Implementar Regresión Ridge**

Copia el pipeline anterior, pero reemplaza `LinearRegression` con `Ridge(alpha=10)`. `alpha` es la fuerza de la penalización.

```python
# Crear el pipeline para Ridge
modelo_ridge = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('model', Ridge(alpha=10))
])
```

```python
# Entrenar el modelo
modelo_ridge.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
# Predicciones
y_train_pred_ridge = modelo_ridge.predict(X_train)
y_test_pred_ridge = modelo_ridge.predict(X_test)

# Cálculo del error (RMSE)
rmse_train_ridge = np.sqrt(mean_squared_error(y_train, y_train_pred_ridge))
rmse_test_ridge = np.sqrt(mean_squared_error(y_test, y_test_pred_ridge))

print(f"RMSE en Entrenamiento (Ridge): ${rmse_train_ridge:,.2f}")
print(f"RMSE en Prueba (Ridge): ${rmse_test_ridge:,.2f}")
```

**Interpreta los resultados:**
El modelo Ridge presenta un RMSE prácticamente idéntico en entrenamiento `($35,204.89) y prueba ($35,206.63)`, lo que indica una excelente capacidad de generalización y ausencia de sobreajuste. La regularización L2 logra controlar eficazmente la complejidad del modelo polinómico, estabilizando los coeficientes y permitiendo que el modelo mantenga un desempeño consistente en datos no observados, lo que lo convierte en una opción robusta y confiable para la predicción de precios de vivienda.


**Implementar Regresión Lasso**

Haz lo mismo, pero ahora con `Lasso(alpha=500, max_iter=10000)`. Lasso necesita un `alpha` más grande (porque la penalización L1 es diferente) y a veces más `max_iter` para converger.

```python
# Crear el pipeline para Lasso
modelo_lasso = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('model', Lasso(alpha=500, max_iter=10000))
])
```

```python
# Entrenar el modelo
modelo_lasso.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
# Predicciones
y_train_pred_lasso = modelo_lasso.predict(X_train)
y_test_pred_lasso = modelo_lasso.predict(X_test)

# Cálculo del error (RMSE)
rmse_train_lasso = np.sqrt(mean_squared_error(y_train, y_train_pred_lasso))
rmse_test_lasso = np.sqrt(mean_squared_error(y_test, y_test_pred_lasso))

print(f"RMSE en Entrenamiento (Lasso): ${rmse_train_lasso:,.2f}")
print(f"RMSE en Prueba (Lasso): ${rmse_test_lasso:,.2f}")
```

**Interpreta los resultados:**

El modelo Lasso muestra un RMSE de` $35,352.92 en entrenamiento y $35,447.71 en prueba`, evidenciando también una buena capacidad de generalización, aunque con un error ligeramente mayor que Ridge. Esto sugiere que, si bien la regularización L1 ayuda a reducir el sobreajuste y simplificar el modelo mediante la selección de variables, en este caso puede estar eliminando información relevante, lo que resulta en una leve pérdida de precisión en las predicciones.

**Selección de Variables con Lasso**

Una de las grandes ventajas de Lasso es que puede eliminar variables. Vamos a ver cuántas de las características polinómicas que creamos (ej. `OverallQual^2`, `GrLivArea^5`, `OverallQual * GrLivArea^4`, etc.) fueron eliminadas.

```python
# Extraer los coeficientes del modelo Lasso
coefs_lasso = modelo_lasso.named_steps['model'].coef_
```

```python
# Extraer los nombres de las características generadas
# Guíate por el siguiente código: feature_names = pipeline_lasso.named_steps['poly_features'].get_feature_names_out(X.columns)
feature_names = modelo_lasso.named_steps['poly'].get_feature_names_out(X.columns)
```

```python
# Contar cuántos coeficientes son exactamente cero
num_zero_coeffs = np.sum(coefs_lasso == 0)
```

```python
# Realizar un print de:
# Número total de características polinómicas generadas
# Número de características eliminadas por Lasso (coef = 0)
# Número de características CONSERVADAS
# Porcentaje de variables eliminadas
# Lista con los nombres de las características conservadas por Lasso (coef != 0)

# Número total de características
num_total = len(coefs_lasso)

# Número de características conservadas
num_nonzero = num_total - num_zero_coeffs

# Porcentaje eliminadas
percent_zero = (num_zero_coeffs / num_total) * 100

# Características conservadas
selected_features = feature_names[coefs_lasso != 0]

print(f"Número total de características polinómicas: {num_total}")
print(f"Número de características eliminadas (coef = 0): {num_zero_coeffs}")
print(f"Número de características CONSERVADAS: {num_nonzero}")
print(f"Porcentaje de variables eliminadas: {percent_zero:.2f}%")

print("\nCaracterísticas conservadas por Lasso:")
print(selected_features)
```

**Interpreta los resultados:**
El modelo Lasso eliminó 15 de las 21 características polinómicas (71.43%), conservando únicamente 6 variables, lo que evidencia una fuerte reducción de la complejidad del modelo. Esto indica que muchas de las combinaciones polinómicas generadas no aportaban valor significativo a la predicción del precio y, por tanto, fueron descartadas automáticamente.

Las variables que se conservaron muestran un patrón claro: la mayoría están relacionadas con OverallQual (calidad de la vivienda), ya sea de forma directa o en interacción con el área habitable (GrLivArea). Esto sugiere que la calidad es un factor estructural clave en la determinación del precio, mientras que el área influye principalmente en combinación con esta. Además, la presencia de términos como GrLivArea^5 indica que el tamaño también tiene un efecto no lineal relevante, aunque más específico.

En conjunto, estos resultados reflejan que Lasso no solo mejora la generalización del modelo, sino que también actúa como un mecanismo de selección de variables, permitiendo identificar de manera automática cuáles son los factores más importantes. Esto hace que el modelo sea más simple, interpretable y enfocado en las relaciones realmente significativas dentro del mercado inmobiliario.
